```python

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Ali', 'Sara', 'John', 'Adeel'],
    'Department': ['IT', 'HR', 'Finance', 'IT']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 5],
    'Salary': [50000, 60000, 55000, 70000]
})

df1, df2
```

```python
pd.merge(df1, df2, on='ID', how='right')
```

```python
pd.merge(df1, df2, on='ID', how='left')
```

```python
pd.merge(df1, df2, on='ID', how='right')
```

```python
pd.merge(df1, df2, on='ID', how='outer')
```

```python

merged = pd.merge(df1, df2, on='ID', how='inner')
merged.groupby('Department')['Salary'].mean()
```

```python
bins = [0, 50000, 60000, 80000]
labels = ['Low', 'Medium', 'High']

merged['Salary_Category'] = pd.cut(merged['Salary'], bins=bins, labels=labels)
merged
```

```python
merged.head()
merged.tail()
merged.columns
merged.index
merged.sort_values(by='Salary')
merged.isnull().sum()
```

```python

merged.to_csv('output.csv', index=True)
```
